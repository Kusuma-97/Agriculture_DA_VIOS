```python
df["Yield_Tonnes_Ha"] = df["Yield_Tonnes_Ha"].fillna(
    df["Production_Tonnes"] / df["Farm_Area_Hectares"]
)
df["Rainfall_mm"] = df["Rainfall_mm"].fillna(df["Rainfall_mm"].median())
df["Soil_Moisture_pct"] = df["Soil_Moisture_pct"].fillna(
    df["Soil_Moisture_pct"].median()
)
df.to_csv("cleaned_agri_data.csv", index=False)
saved_df = pd.read_csv("cleaned_agri_data.csv")
df.info()
```

```python
cat_cols = ["State", "Crop", "Season", "Irrigation_Method"]
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()
for ax, col in zip(axes, cat_cols):
    order = df[col].value_counts().index
    sns.countplot(data=df, y=col, order=order, ax=ax)
    ax.set_title(f"Farm Count by {col}")
    ax.set_xlabel("Count")
    ax.set_ylabel("")
plt.tight_layout()
plt.show()
```

```python
num_cols = ["Farm_Area_Hectares", "Rainfall_mm", "Avg_Temperature_C", "Yield_Tonnes_Ha",
            "Production_Tonnes", "Profit_INR", "Water_Used_m3", "Disease_Pest_Risk_pct"]

fig, axes = plt.subplots(4, 2, figsize=(14, 18))
axes = axes.flatten()
for ax, col in zip(axes, num_cols):
    sns.histplot(df[col], kde=True, ax=ax, color="teal")
    ax.set_title(f"Distribution of {col}")
    ax.axvline(df[col].mean(), color="red", linestyle="--", linewidth=1, label="Mean")
    ax.axvline(df[col].median(), color="orange", linestyle="--", linewidth=1, label="Median")
    ax.legend(fontsize=8)
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 4, figsize=(18, 8))
axes = axes.flatten()
for ax, col in zip(axes, num_cols):
    sns.boxplot(y=df[col], ax=ax, color="lightseagreen")
    ax.set_title(col, fontsize=10)
plt.tight_layout()
plt.show()
```

```python
#KEY INSIGHTS
df["Profit_per_Hectare"] = df["Profit_INR"] / df["Farm_Area_Hectares"]
df["Is_Profitable"] = df["Profit_INR"] > 0
df["Profit_Margin_pct"] = (df["Profit_INR"] / df["Revenue_INR"]) * 100
corr = df.corr(numeric_only=True)

# Aggregations
best_crop_yield = df.groupby("Crop")["Yield_Tonnes_Ha"].mean().idxmax()
best_crop_profit = df.groupby("Crop")["Profit_INR"].mean().idxmax()
best_state_profit = df.groupby("State")["Profit_per_Hectare"].mean().idxmax()
best_irrigation = df.groupby("Irrigation_Method")["Water_Efficiency_t_per_1000m3"].mean().idxmax()
riskiest_crop = df.groupby("Crop")["Disease_Pest_Risk_pct"].mean().idxmax()
best_season_yield = df.groupby("Season")["Yield_Tonnes_Ha"].mean().idxmax()
print("=" * 60)
print("KEY INSIGHTS")
print("=" * 60)
print(f"• Total farms analyzed:            {len(df):,}")
print(f"• Overall % profitable farms:      {df['Is_Profitable'].mean()*100:.1f}%")
print(f"• Highest average yield crop:      {best_crop_yield}")
print(f"• Highest average profit crop:     {best_crop_profit}")
print(f"• Best season for yield:           {best_season_yield}")
print(f"• Most profitable state (per ha):  {best_state_profit}")
print(f"• Most water-efficient irrigation: {best_irrigation}")
print(f"• Highest disease/pest risk crop:  {riskiest_crop}")
print(f"• Avg profit margin:               {df['Profit_Margin_pct'].mean():.1f}%")
print(f"• Correlation (Rainfall, Yield):   {corr.loc['Rainfall_mm','Yield_Tonnes_Ha']:.2f}")
print(f"• Correlation (Fertilizer, Yield): {corr.loc['Fertilizer_kg_ha','Yield_Tonnes_Ha']:.2f}")
print(f"• Correlation (Disease Risk, Yield): {df['Disease_Pest_Risk_pct'].corr(df['Yield_Tonnes_Ha']):.2f}")
print("=" * 60)
```

```python
df["Profit_Margin_pct"] = (df["Profit_INR"] / df["Revenue_INR"]) * 100
df["Is_Loss_Making"] = df["Profit_INR"] < 0

# 3. Comprehensive Seasonal Aggregations
seasonal_analysis = df.groupby("Season").agg(
    Total_Farms=("Farm_ID", "count"),
    Mean_Yield_t_ha=("Yield_Tonnes_Ha", "mean"),
    Std_Yield_t_ha=("Yield_Tonnes_Ha", "std"),
    Mean_Production_t=("Production_Tonnes", "mean"),
    Mean_Rainfall_mm=("Rainfall_mm", "mean"),
    Mean_Temp_C=("Avg_Temperature_C", "mean"),
    Mean_Humidity_pct=("Humidity_pct", "mean"),
    Mean_Moisture_pct=("Soil_Moisture_pct", "mean"),
    Pest_Risk_pct=("Disease_Pest_Risk_pct", "mean"),
    Mean_Water_Used_m3=("Water_Used_m3", "mean"),
    Water_Efficiency_t_1000m3=("Water_Efficiency_t_per_1000m3", "mean"),
    Fertilizer_kg_ha=("Fertilizer_kg_ha", "mean"),
    Pesticide_L_ha=("Pesticide_Litre_ha", "mean"),
    Mean_Total_Cost_INR=("Total_Cost_INR", "mean"),
    Mean_Revenue_INR=("Revenue_INR", "mean"),
    Mean_Profit_INR=("Profit_INR", "mean"),
    Loss_Making_Farm_pct=("Is_Loss_Making", lambda x: x.mean() * 100),
)

print("=== SEASONAL AGRICULTURAL PERFORMANCE SUMMARY ===")
print(seasonal_analysis.T)

# 4. Specific Cross-Tabulations (Crops & Irrigation across Seasons)
print("\n=== CROP DISTRIBUTION ACROSS SEASONS (%) ===")
print(pd.crosstab(df["Season"], df["Crop"], normalize="index") * 100)

print("\n=== IRRIGATION METHOD DISTRIBUTION ACROSS SEASONS (%) ===")
print(
    pd.crosstab(df["Season"], df["Irrigation_Method"], normalize="index") * 100
)

# 5. Regional Consistency (Mean Profit per State across Seasons)
print("\n=== MEAN PROFIT BY STATE & SEASON (INR) ===")
print(
    df.groupby(["State", "Season"])["Profit_INR"]
    .mean()
    .unstack()
    .round(2)
)
```
